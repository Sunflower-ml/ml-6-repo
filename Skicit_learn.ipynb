```python
import pandas as pd

# 1. Create our training dataset
data = [
    {"years_exp": 1, "salary": 40000},
    {"years_exp": 2, "salary": 50000},
    {"years_exp": 3, "salary": 60000},
    {"years_exp": 4, "salary": 70000},
    {"years_exp": 5, "salary": 80000}
]

df = pd.DataFrame(data)

# 2. Split into Features (X) and Target (y)
# Note: X uses double brackets [[ ]] because features are a 2D table!
X = df[["years_exp"]]
y = df["salary"]

print("Features (X):")
print(X)
print("\nTarget (y):")
print(y)
```

```python
from sklearn.linear_model import LinearRegression

# 1. Create the empty Machine Learning model
model = LinearRegression()

# 2. Train (.fit) the model on our data (X and y)
model.fit(X, y)

# 3. Predict the salary for someone with 6 years of experience!
# (Notice we pass a small DataFrame with the exact same column name)
new_person = pd.DataFrame([{"years_exp": 6}])
prediction = model.predict(new_person)

print(f"Predicted salary for 6 years experience: ${prediction[0]}")
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

houses = pd.DataFrame([
    {"rooms": 1, "price": 100000},
    {"rooms": 2, "price": 150000},
    {"rooms": 3, "price": 200000},
    {"rooms": 4, "price": 250000}
])

X = houses[["rooms"]]
y = houses["price"]

house_model = LinearRegression()
house_model.fit(X,y)

new_house  = pd.DataFrame([{"rooms": 5}])
predicted_price = house_model.predict(new_house)
print(f"predicted price{predicted_price}")
```

```python

from sklearn.model_selection import train_test_split

# 1. A slightly larger dataset (8 rows)
df_salaries = pd.DataFrame([
    {"exp": 1, "salary": 40000},
    {"exp": 2, "salary": 51000},
    {"exp": 3, "salary": 59000},
    {"exp": 4, "salary": 70000},
    {"exp": 5, "salary": 81000},
    {"exp": 6, "salary": 89000},
    {"exp": 7, "salary": 100000},
    {"exp": 8, "salary": 110000}
])

X = df_salaries[["exp"]]
y = df_salaries["salary"]

# 2. Automatically split 75% for training, 25% for testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# 3. Train ONLY on the training set!
model = LinearRegression()
model.fit(X_train, y_train)

# 4. Evaluate accuracy on the hidden test set! (1.0 is a perfect 100% score)
r2_score = model.score(X_test, y_test)
print(f"Model test score: {round(r2_score * 100, 2)}%")

# Let's print the 4 boxes created by train_test_split!
print("--- HOMEWORK (75% = 6 rows) ---")
print(X_train)

print("\n--- FINAL EXAM QUESTIONS (25% = 2 rows) ---")
print(X_test)

print("\n--- What the model GUESSED on the exam vs REAL answers ---")
exam_guesses = model.predict(X_test)
print(f"Model's Guesses: {exam_guesses}")
print(f"Real Answers:    {y_test.values}")
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

cars = pd.DataFrame([
    {"age": 1, "km": 10, "price": 28000},
    {"age": 2, "km": 25, "price": 25000},
    {"age": 3, "km": 40, "price": 22000},
    {"age": 4, "km": 50, "price": 19500},
    {"age": 5, "km": 70, "price": 16000},
    {"age": 6, "km": 85, "price": 13000},
    {"age": 7, "km": 100, "price": 10000},
    {"age": 8, "km": 120, "price": 6500}
])

X = cars[["age", "km"]]
y = cars["price"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)

print(f"score {round(model.score(X_test, y_test) * 100, 2)}")
print(f"predict {model.predict(pd.DataFrame([{"age": 3, "km": 30}]))}")
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split

servers = pd.DataFrame([
    {"cpu": 20, "ram": 30, "crash": 0},
    {"cpu": 25, "ram": 35, "crash": 0},
    {"cpu": 30, "ram": 40, "crash": 0},
    {"cpu": 40, "ram": 45, "crash": 0},
    {"cpu": 85, "ram": 90, "crash": 1},
    {"cpu": 90, "ram": 88, "crash": 1},
    {"cpu": 95, "ram": 92, "crash": 1},
    {"cpu": 98, "ram": 96, "crash": 1}
])

X = servers[["cpu", "ram"]]
y = servers["crash"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.25, random_state=42)

model = DecisionTreeClassifier()
model.fit(X_train, y_train)

print(f"score {model.score(X_test,y_test) *100}")
print(f"predict {model.predict(pd.DataFrame([{"cpu": 92, "ram": 89}]))}")
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split

# links = number of links in email, words = word count, spam = 1 (Spam) or 0 (Safe)
emails = pd.DataFrame([
    {"links": 0, "words": 150, "spam": 0},
    {"links": 1, "words": 200, "spam": 0},
    {"links": 1, "words": 120, "spam": 0},
    {"links": 2, "words": 300, "spam": 0},
    {"links": 8, "words": 50,  "spam": 1},
    {"links": 10, "words": 40, "spam": 1},
    {"links": 12, "words": 60, "spam": 1},
    {"links": 1, "words": 180, "spam": 1}  # <-- Tricky row! Looks like a normal email, but is Spam!
])

X = emails[["links", "words"]]
y = emails["spam"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state= 42)

tree = RandomForestClassifier(random_state = 42)
tree.fit(X_train, y_train)

print(f"score {tree.score(X_test, y_test)* 100}")
print(f"predict {tree.predict(pd.DataFrame([{"links": 9, "words": 45}]))}")
```
