```python
def add_bonus(score):
    new_score = score + 5
    return new_score
final_result = 88
print(add_bonus(final_result))
```

```python
def evaluate_models(score_list):
  for score in score_list:
    if score >= 85:
      print(f"{score}%: Passed")
    else :
      print(f"{score}%: Failed")
my_models = [75, 88, 92, 60]
evaluate_models(my_models)
```
