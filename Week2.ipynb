```python
import pandas as pd

# 1. Start with a List of Dictionaries (just like earlier today)
data = [
    {"model": "LinearRegression", "framework": "Scikit-learn", "accuracy": 78.5},
    {"model": "RandomForest", "framework": "Scikit-learn", "accuracy": 89.0},
    {"model": "ResNet", "framework": "PyTorch", "accuracy": 94.2},
    {"model": "MobileNet", "framework": "TensorFlow", "accuracy": 91.8}
]

# 2. Convert it into a Pandas DataFrame (table)
df = pd.DataFrame(data)

# 3. Display the table!
print(df)
```

```python
# Grab a single column by its name:
print("All accuracies:")
print(df["accuracy"])

# Calculate the average of a column:
print(f"\nAverage Accuracy: {df['accuracy'].mean()}%")

# Add 2 bonus points to the entire 'accuracy' column at once:
df["curved_accuracy"] = df["accuracy"] + 2
print("\nUpdated Table with new column:")
print(df)
```

```python
import numpy as np
import pandas as pd

# 1. A dataset of houses where the 3rd house is missing its 'age' (np.nan)
house_data = [
    {"price": 250000, "rooms": 3, "age": 10},
    {"price": 320000, "rooms": 4, "age": 5},
    {"price": 180000, "rooms": 2, "age": np.nan},  # Missing value!
    {"price": 400000, "rooms": 5, "age": 20}
]

df_houses = pd.DataFrame(house_data)
print("Before cleaning:")
print(df_houses)

# 2. Fill the missing 'age' with the median (middle) age of the column:
median_age = df_houses["age"].median()
df_houses["age"] = df_houses["age"].fillna(median_age)

print("\nAfter cleaning (NaN replaced with median):")
print(df_houses)
```

```python
messy_models = pd.DataFrame([
    {"model": "Model_A", "accuracy": 80.0},
    {"model": "Model_B", "accuracy": np.nan},
    {"model": "Model_C", "accuracy": 90.0}
])

print(messy_models["accuracy"].mean())

NoNr = messy_models["accuracy"].median()
messy_models['accuracy'] = messy_models['accuracy'].fillna(NoNr)

print(messy_models)
```
